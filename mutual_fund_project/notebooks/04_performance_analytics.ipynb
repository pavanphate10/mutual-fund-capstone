# Performance Analytics
This notebook summarizes return, risk, regression, drawdown, scorecard, and benchmark-comparison analytics for the 40 mutual fund schemes in the capstone dataset.

## Data Coverage and Assumptions
- Input files came from `data/processed/nav_history.csv`, `fund_master.csv`, and `benchmark_indices.csv`.
- NAV and benchmark history spans **2022-01-03 to 2026-05-29**.
- Daily returns use `nav_t / nav_t-1 - 1`.
- Sharpe and Sortino use a **6.5% annual risk-free rate** with 252 trading days.
- Alpha and beta come from `scipy.stats.linregress` of fund daily returns against `NIFTY100` daily returns.
- Because the available history is under five years, trailing **5-year CAGR is unavailable** and remains `NaN` for all schemes.
