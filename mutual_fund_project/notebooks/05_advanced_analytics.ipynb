# Advanced Analytics
This notebook consolidates tail-risk, rolling risk-adjusted return, investor behavior, SIP continuity, portfolio concentration, and recommendation outputs for the 40 mutual fund schemes.

## Methodology
- Historical VaR (95%) uses the 5th percentile of daily returns by fund.
- CVaR (95%) uses the mean of daily returns that fall at or below the VaR threshold.
- Rolling Sharpe ratio uses `returns.rolling(90).mean() / returns.rolling(90).std() * sqrt(252)`.
- The 5 key funds in the chart are the top 5 schemes by AUM, with Sharpe ratio as a tiebreaker.
- SIP continuity flags investors as at-risk when average SIP gaps exceed 35 days after at least 6 SIP transactions.
- HHI is calculated as the sum of squared portfolio weights in decimal form for disclosed equity holdings.
